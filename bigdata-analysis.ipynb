```python
import os
import shutil
import pandas as pd

# 1. Create a local working directory
project_dir = "/kaggle/working/fever_project"
wiki_dir = os.path.join(project_dir, "wiki_pages")
os.makedirs(wiki_dir, exist_ok=True)

# 2. Copy your uploaded feverrr files (train & dev jsonl) into the project folder
source_feverrr = "/kaggle/input/datasets/niyatiiiiiiiii/feverrr"
for filename in os.listdir(source_feverrr):
    src = os.path.join(source_feverrr, filename)
    dst = os.path.join(project_dir, filename)
    if os.path.isfile(src):
        shutil.copy(src, dst)

print(f"✔ Copied feverrr files to {project_dir}")

# 3. Export a sample or the full wiki corpus into the project folder if needed
# (Saving the first 10,000 wiki items as a local jsonl for quick offline access)
wiki_local_path = os.path.join(wiki_dir, "wiki_sample.jsonl")
print("Exporting wiki sample to local folder...")

with open(wiki_local_path, "w", encoding="utf-8") as f:
    for i in range(min(10000, len(wiki_corpus))):
        f.write(json.dumps(wiki_corpus[i]) + "\n")

print(f"✔ Local fever_project folder is ready at: {project_dir}")
print("Contents:", os.listdir(project_dir))
```
